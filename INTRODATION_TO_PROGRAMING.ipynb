```python
#01 Py2th0on3
text = input("Enter a string: ")
word = ""
sum_numbers = 0
for char in text:
    if char.isalpha():
        word = word + char
    elif char.isdigit():
        sum_numbers = sum_numbers + int(char)
print('Word: "' + word + '", Sum:', sum_numbers)
```

```python
#02 Guest@123
password = input("Enter password: ")
if 8 <= len(password) <= 16:
    has_upper = False
    has_special = False
    for char in password:
        if char.isupper():
            has_upper = True
        elif char in "@#$!%^&*":
            has_special = True
    if has_upper and has_special:
        print("Password is Secure.")
    else:
        print("Invalid password")
else:
    print("Invalid password")
```

```python
#03 
price = float(input("Price: "))
member = input("Member (Yes/No): ")
if price < 50:
    final_price = price
elif 50 <= price <= 100:
    final_price = price - (price * 0.10)
else:
    final_price = price - (price * 0.20)
if member == "Yes":
    final_price = final_price - (final_price * 0.05)
print("Final Total: $" + str(final_price))
```

```python
#05 {'Sam': [95, 90, 92]}
data_str = input("Enter dictionary: ")
data = eval(data_str)
for name in data:
    scores = data[name]
    total = 0
    for score in scores:
        total = total + score
    avg = total / len(scores)
    if avg > 90:
        print(name + ": Average", avg, "- Honor Roll")
```

```python
#07
num = input("Enter number: ")
while len(num) > 1:
    total = 0
    for digit in num:
        total = total + int(digit)
    num = str(total)
print("Digital Root:", num)
```

```python
#08 [1, 1, 2, 3, 3, 3, 4]
lst_str = input("Enter list of numbers: ")
lst = eval(lst_str)
result = []
counted = []
for num in lst:
    if num not in counted:
        count = 0
        for n in lst:
            if n == num:
                count = count + 1
        result.append((num, count))
        counted.append(num)
print(result)
```

```python
#09
n = int(input("Enter n: "))
factorial = 1
for i in range(1, n + 1):
    factorial = factorial * i
total_sum = 0
for i in range(1, n + 1):
    total_sum = total_sum + i
if factorial % total_sum == 0:
    print("Factorial is " + str(factorial) + ", Sum is " + str(total_sum) + ". " + str(factorial) + " is divisible by " + str(total_sum) + ".")
else:
    print("Factorial is " + str(factorial) + ", Sum is " + str(total_sum) + ". " + str(factorial) + " is not divisible by " + str(total_sum) + ".")
```

```python
#10
store = {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}
total = 0.0
while True:
    action = input("Enter Add 'item' or 'Done': ")
    if action == "Done":
        break
    parts = action.split(" ")
    if len(parts) == 2 and parts[0] == "Add":
        item = parts[1].replace("'", "")
        if item in store:
            total = total + store[item]
        else:
            print("Error")
print("Total Bill: $" + str(total))
```

```python
#11 -- Race car
sentence = input("Enter sentence: ")
clean = ""
for char in sentence:
    if char != " ":
        clean = clean + char.lower()
reverse = ""
for i in range(len(clean) - 1, -1, -1):
    reverse = reverse + clean[i]
if clean == reverse:
    print("True")
else:
    print("False")
```

```python
#13 -- [1, 5, 2, 6, 3]
lst_str = input("Enter list: ")
lst = eval(lst_str)
peaks = []
for i in range(1, len(lst) - 1):
    if lst[i] > lst[i-1] and lst[i] > lst[i+1]:
        peaks.append(lst[i])
print("Peaks:", peaks)
```

```python
#14 {'A': 1, 'B': 2, 'C': 1}
d_str = input("Enter dictionary: ")
d = eval(d_str)
flipped = {}
for key in d:
    val = d[key]
    if val in flipped:
        flipped[val].append(key)
    else:
        flipped[val] = [key]
print(flipped)
```

```python
#15 banana
text = input("Enter word: ")
vowels = {}
for i in range(len(text)):
    if text[i] in "aeiouAEIOU":
        char = text[i].lower()
        if char in vowels:
            vowels[char].append(i)
        else:
            vowels[char] = [i]
print(vowels)
```
